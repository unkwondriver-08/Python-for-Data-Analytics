In this tutorial we will work again in the time series data where we will work with date range function where we will update the date  in a csv where we dont have dates

```python
import pandas as pd
```

```python
# create the date range
df = pd.read_excel("C:\\MY ALL\\Analytics Study\\Python\\Pandas\\panda_saving\\appl.xlsx")
df.head()
```

```python
rng = pd.date_range(start = '2024-01-01', end = '2024-01-23', freq = 'B') # B = Business Days
rng
```

```python
# set the date range into the dataframe
df.set_index(rng, inplace=True)
```

```python
%matplotlib inline
df.Close.plot()
```

```python
df.loc['2024-01-01':'2024-01-10']
```

```python
# suppose we want to modify our dataframe to include the dates which are weekends etc we can do so by using asfreq

df.asfreq('D', method='pad') # D = Daily, method = pad means fill the missing values with the last available value
```

```python
# you can do for hourly etc as well by changing the freq parameter to 'H' for hourly, 'W' for weekly, etc.
df.asfreq('h', method='pad') # h = Hourly
```

```python
# suppose while creating the date rang, i knw the start date but not the end date but i knw the frequency of the data, i can use periods parameter to create the date range
rng2 = pd.date_range(start = '2024-01-01', periods = 17, freq = 'B') # B = Business Days
```

```python
df.set_index(rng2, inplace=True)
df.head()
```
