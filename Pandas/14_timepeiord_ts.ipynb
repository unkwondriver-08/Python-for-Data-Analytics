```python
# let us create a period

import pandas as pd
y = pd.Period('2026')
y

# Y - means annual and dec means ending in december. So this period represents the year 2026, which starts on January 1, 2026 and ends on December 31, 2026. The frequency of this period is annual, which means that it represents a single year.
```

Period Index - peiods can be used as a index in Dataframes

Think of period_range() as:

"Start at this period and keep moving by one quarter until you reach the ending period."

```python
idx = pd.period_range('2026-01', '2026-12', freq='Q')
idx
```

```python
idx = pd.period_range('2016', '2026', freq='Q')
idx

# here fiscal year is from jan to dec and  hence 2016 Q1 is first entrance
```

```python
idx = pd.period_range('2016', '2026', freq='Q-JAN')
idx

# here fiscal year is from feb to dec and  hence it interprets the year 2016 as the fiscal year ending in January 2016. so the
# first quarter of the fiscal year 2016 is from February 1, 2015 to April 30, 2015, and the last quarter of the fiscal year 2026 is from November 1, 2025 to January 31, 2026.
# the january quarter is starting of our series because we have started from 2016 (jan) which is intself the q4
```

```python
idx.start_time
```

```python
import pandas as pd

df = pd.DataFrame({
    "Line Item": ["Revenue", "Expense", "Profit"],
    "2025Q1": [100000, 70000, 30000],
    "2025Q2": [120000, 80000, 40000],
    "2025Q3": [130000, 85000, 45000],
    "2025Q4": [150000, 90000, 60000],
    "2026Q1": [170000, 100000, 70000]
})

df
```

```python
df.set_index('Line Item', inplace=True)
df
```

```python
df= df.T
df
```

```python
df.index 
# here the index is a string, we can convert it to a period index using the pd.PeriodIndex() function. We can also specify the frequency of the period index using the freq parameter. In this case, we want to use a quarterly frequency, so we set freq='Q'.
```

```python
df.index = pd.PeriodIndex(df.index, freq='Q')
df.index
```

```python
df['start_date']= df.index.map(lambda x: x.start_time)
df
```

```python
df['end_date'] = df.index.map(lambda x: x.end_time.date())
df
```
