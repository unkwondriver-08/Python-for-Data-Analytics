merge acts as Join here we can use the how act as key allowing inner, outer, left

```python
import pandas as pd
India_weather = pd.DataFrame( {
    "city": ['Bihar','Delhi', 'Mumbai'],
    "temprature":[ 35, 40, 30]})
India_weather
```

```python
India_humidity = pd.DataFrame( {
    "city": ['Bihar','Delhi', 'Mumbai'],
    "Humidity":[ 65,40, 90],})
India_humidity
```

```python

pd.merge(India_weather, India_humidity, on ='city', how= 'outer')
```

```python
pd.merge(India_weather, India_humidity, on ='city', how= 'inner')
```

```python
#  you can use indicator column to tell which data come from which tabloe
pd.merge(India_weather, India_humidity, on ='city', how= 'inner', indicator= True)
```
