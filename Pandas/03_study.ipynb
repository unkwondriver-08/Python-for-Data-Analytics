we will modify and edit missing data or wrong values using Replace function


```python
import pandas as pd
import numpy as np
df = pd.read_csv("C:\\MY ALL\\Analytics Study\\Python\\Pandas\\panda_saving\\date_max_min_price.csv")
df
```

```python
new_df = df.replace(-9999, np.nan)
new_df
```

```python
new_df = df.replace([-9999, 19.0], np.nan)
new_df
```

but sometimes replacing the  value is not the corrcet method as some row could have value which are correct for that case.
eg - replacing -1 will remove all -1 but -1 can be significant in some columns\
we solve this problem using dictionary



```python
new_df = df.replace({
    'Max': -9999,
    'Min': 19.0,
    'Events': 'No Event'
}, np.nan)
new_df
```

replacing  with mapping


```python
new_df = df.replace({
     -9999:np.nan,
    19.0: np.nan,
    'No Event': 'very good'
})
new_df
```

using regexp to convert the expersion values into correct values on specific columns



```python
new_df = df.replace({
        'Events': '[0-9]+'
    },'', regex=True
)
new_df
```

```python
import pandas as pd

data = {
    "Student": ["Rahul", "Priya", "Aman", "Sneha", "Rohit", "Ananya", "Vikas", "Neha"],
    "Grade": [ "Good", "Exceptional", "Average","Bad","Good","Exceptional","Poor","Average"]
}

df = pd.DataFrame(data)

df
```

```python
new_df = df.replace(
    {
        "Grade": {
            "Good": "A",
            "Exceptional": "A+",
            "Average": "B",
            "Bad": "C",
            "Poor": "D"
        }
    }
)
new_df
```
