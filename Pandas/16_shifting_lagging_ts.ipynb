it helps in calculating the change in certain amount of days


```python
import pandas as pd

df = pd.DataFrame({
    'Price': [178.88, 175.88, 173.88, 172.88, 171.88, 171.88]
}, index=pd.to_datetime([
    '2025-01-01 09:30:00',
    '2025-01-02 09:30:00',
    '2025-01-03 09:30:00',
    '2025-01-04 09:30:00',
    '2025-01-05 09:30:00',
    '2025-01-06 09:30:00'
]))

df.index.name = 'Date'

df
```

```python
df.shift(2)
# will shift the data point in the right hand side
```

```python
# left shifting

df.shift(-2)
```

```python
df['previous_day_price']= df['Price'].shift(1)
df
```

```python
df['1 day price_change'] = df['Price'] - df['previous_day_price']
df
```

```python
df['5day //% return '] = (df['Price']- df['Price'].shift(5))*100/df['Price'].shift(5)
df
```
