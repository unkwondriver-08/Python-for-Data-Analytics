```python
import pandas as pd

df = pd.DataFrame({
    'Price': [78.88, 75.88, 73.88, 72.88, 71.88, 71.88]
}, index=pd.to_datetime([
    '2025-01-01 09:30:00',
    '2025-01-02 09:30:00',
    '2025-01-03 09:30:00',
    '2025-01-04 09:30:00',
    '2025-01-05 09:30:00',
    '2025-01-06 09:30:00'
]))

df.index.name = 'Date'

df
```

```python
df.index
```

Two types of DateTime objects in python

1. Naive (No timezone awareness)
2. Time zone aware datetime

```python
df= df.tz_localize('America/New_York')
```

```python
df
```

```python
# convert the timezone to UTC
df_utc = df.tz_convert('UTC')
df_utc
```

```python
df_india = df.tz_convert('Asia/Kolkata')
df_india
```

```python
df.index= df.index.tz_convert('Asia/Kolkata')
df

# does the same thing as above
```

```python
rng = pd.date_range('2025-01-01', periods=6, freq='h')
rng
# by default it is naive, meaning it has no timezone information.
#  We can localize it to a specific timezone using the tz_localize method:
# datetime64[us] means microsecond precision,
```

```python
rng = pd.date_range('2025-01-01', periods=6, freq='h', tz='Asia/Kolkata')
rng
```

```python
s = pd.Series(range(len(rng)), index=rng)
s
```

```python
b= s.tz_convert('America/New_York')
b
```

tz_localize() = tell Pandas what timezone the datetime is already in. So for the first time we will use this only to let DF knw its Timezone
tz_convert() = change an already timezone-aware datetime to another timezone.

```python
m = s.tz_convert('Asia/Shanghai')
m
```

```python
b+m

# it will first convert both the b and m into utc and then do the arithemetic
```
