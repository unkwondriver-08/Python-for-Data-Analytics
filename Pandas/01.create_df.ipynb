using Convertor to handle messy data
here we define a function and then use that function in the data frame line eg


```python
import pandas as pd
df = pd.read_csv('C:\\MY ALL\\Analytics Study\\Python\\archive\\modified_weather.csv')
df.head(5)
```

```python
import pandas as pd
def convert_to_min(cell):
    if cell == "n.a":
        return None
    return cell

def convert_to_max(cell):
    if cell == "Not available":
        return None
    return cell

df = pd.read_csv('C:\\MY ALL\\Analytics Study\\Python\\archive\\modified_weather.csv',
                 converters={"MinTemp": convert_to_min, 
                             "MaxTemp": convert_to_max}  )
df.head(5)
```

```python
df.to_excel('C:\\MY ALL\\Analytics Study\\Python\\Pandas\\panda_saving\\modified_weather2.xlsx', startrow=2, startcol=3, index=False) # index=False means we don't want to write row numbers 
```

save two dfs in same xcel but different sheets


```python
import pandas as pd

data1 = {
    "Name": ["A", "B", "C"],
    "Age": [20, 21, 22],
    "Marks": [80, 90, 85]
}

data2 = {
    "Name": ["D", "E", "F"],
    "Age": [23, 24, 25],
    "Marks": [75, 88, 92]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

print(df1)
print(df2)
```

```python
with pd.ExcelWriter('C:\\MY ALL\\Analytics Study\\Python\\Pandas\\panda_saving\\multiple_sheets.xlsx') as writer:
    df1.to_excel(writer, sheet_name='Sheet1', index=False)
    df2.to_excel(writer, sheet_name='Sheet2', index=False)
```
